```python
# 1: Import library
import pandas as pd
import matplotlib.pyplot as plt

# : Upload file excel ke Colab
from google.colab import files
uploaded = files.upload()

# 3: Baca data excel
import pandas as pd
path = "naldi.xlsx"
dataraw = pd.read_excel(path)
print(dataraw)
```

```python
# 4: Tabel Frekuensi Grade
import pandas as pd
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print(datafrq)
```

```python
# 5: Grafik Garis
import matplotlib.pyplot as plt
datafrq.plot(color="darkorange", marker="o", linewidth=2)
plt.title("Grafik Garis Frekuensi Grade - Tri Naldi Syaputra Siregar")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")
plt.grid(alpha=0.3)
plt.show()
```

```python
# 6: Grafik Batang
import matplotlib.pyplot as plt
datafrq.plot(kind='bar', color="teal")
plt.title("Grafik Batang Frekuensi Grade - Tri Naldi Syaputra Siregar")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")
plt.xticks(rotation=0)
plt.show()
```

```python
# 7: Pie Chart
import matplotlib.pyplot as plt
datafrq.plot(
    kind='pie',
    y='Frekuensi',
    autopct='%1.0f%%',
    legend=False,
    colors=["#4C72B0", "#DD8452", "#55A868", "#C44E52", "#8172B2", "#937860"],
)
plt.title("Persentase Grade - Tri Naldi Syaputra Siregar")
plt.ylabel("")
plt.show()
```

```python
# 8: Statistika Deskriptif
import pandas as pd
dataraw["Nilai"] = pd.to_numeric(dataraw["Nilai"], errors='coerce')
dt = dataraw["Nilai"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print(stats)

# (tambahan): Kuartil & deteksi outlier dengan IQR
Q1 = dt.quantile(0.25)
Q3 = dt.quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR
outlier = dt[(dt < batas_bawah) | (dt > batas_atas)]
print("Q1 =", Q1, " Q3 =", Q3, " IQR =", IQR)
print("Batas bawah =", batas_bawah, " Batas atas =", batas_atas)
print("Outlier ditemukan:", outlier.tolist())
```
